```python
import networkx as nx
import scipy as sp
import pandas as pd
from collections import Counter
import matplotlib.pyplot as plt
import os
if not "src" in os.listdir(): # I use this to change my directory to access the src and data folders from in here
    os.chdir("..") 

from src.reading_in import read_in_proteins


from STM_alg.stm import compute_ppi_signals, build_preclusters, merge_clusters, run_stm_algorithm, evaluate_network_modules

FILEPATH = "data/raw/protein.links.v12.0.txt.gz"
G = read_in_proteins(include_essential_proteins=False, connected=True)
```

```python
print(f"Network Stats: {G.number_of_nodes()} Nodes, {G.number_of_edges()} Edges")
print("Executing Phase 1: Computing Signal Transduction Matrices...")
signals = compute_ppi_signals(G)
print("Executing Phase 2: Generating Core Pre-clusters...")
preclusters = build_preclusters(G, signals)
```

```python
print(f"Executing Phase 3: Merging overlapping structures (Threshold >= {1.0})...")
final_modules = merge_clusters(G, preclusters, threshold=0.15)
```

```python
sizes = [len(m) for m in final_modules]
print(f"Max size: {max(sizes)}")
print(f"Number of micro-modules (<4 proteins): {sum(1 for s in sizes if s < 4)}")
```
